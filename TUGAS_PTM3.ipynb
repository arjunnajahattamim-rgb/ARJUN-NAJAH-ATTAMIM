```python
# Program input-output sederhana di Python

nama = input("Masukkan nama Anda: ")
umur = int(input("Masukkan umur Anda: "))
hobi = input("Masukkan hobi Anda: ")

print("\n--- Data yang Anda Masukkan ---")
print("Nama  :", nama)
print("Umur  :", umur)
print("Hobi  :", hobi)
```

```python
%%writefile BUKANBAYITABUNG.cpp
#include <iostream>

int main() {
    char nama[50];
    char hobi[50];
    int umur;

    std::cout << "Masukkan nama Anda: ";
    std::cin.getline(nama, 50);

    std::cout << "Masukkan umur Anda: ";
    std::cin >> umur;
    std::cin.ignore(); // Buang newline tersisa

    std::cout << "Masukkan hobi Anda: ";
    std::cin.getline(hobi, 50);

    std::cout << "\nData yang Anda masukkan:\n";
    std::cout << "Nama : " << nama << "\n";
    std::cout << "Umur : " << umur << "\n";
    std::cout << "Hobi : " << hobi << "\n";

    return 0;
}
```

```python
%%shell
g++ BUKANBAYITABUNG.cpp -o main
./main
```
