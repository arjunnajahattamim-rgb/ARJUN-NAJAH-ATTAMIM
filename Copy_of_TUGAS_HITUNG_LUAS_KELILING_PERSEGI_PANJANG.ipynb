```python
# Program Penjumlahan Dua Bilangan
# Meminta input dari pengguna
angka1 = float(input("Masukkan angka pertama: "))
angka2 = float(input("Masukkan angka kedua: "))

# Menjumlahkan kedua angka
hasil = angka1 + angka2

# Menampilkan hasil
print("Hasil penjumlahan:", hasil)
```

```python
# Program Mengecek Cukup Umur
# Meminta input nama dan umur
nama = input("Masukkan nama Anda: ")
umur = int(input("Masukkan umur Anda: "))

# Mengecek apakah cukup umur
if umur >= 18:
    print(nama, "sudah cukup umur (>= 18 tahun).")
else:
    print(nama, "belum cukup umur (< 18 tahun).")
```

```python
# Program menghitung luas dan keliling persegi panjang

# Input dari pengguna
panjang = float(input("Masukkan panjang persegi panjang: "))
lebar = float(input("Masukkan lebar persegi panjang: "))

# Menghitung luas dan keliling
luas = panjang * lebar
keliling = 2 * (panjang + lebar)

# Menampilkan hasil
print("\nHasil Perhitungan:")
print("Luas persegi panjang    =", luas)
print("Keliling persegi panjang =", keliling)
```
