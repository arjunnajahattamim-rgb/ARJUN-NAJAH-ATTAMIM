```python
%%writefile main.cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    string nama, hobi;
    int umur;

    cout << "Masukkan nama Anda: ";
    getline(cin, nama);

    cout << "Masukkan umur Anda: ";
    cin >> umur;
    cin.ignore(); // membersihkan newline

    cout << "Masukkan hobi Anda: ";
    getline(cin, hobi);

    cout << "\n--- Data yang Anda Masukkan ---" << endl;
    cout << "Nama  : " << nama << endl;
    cout << "Umur  : " << umur << endl;
    cout << "Hobi  : " << hobi << endl;

    return 0;
}
```

```python
%%shell
g++ main.cpp -o main
./main
```

```python
# Program input-output sederhana di Python

# Meminta input dari pengguna
nama = input("Masukkan nama Anda: ")
umur = int(input("Masukkan umur Anda: "))
hobi = input("Masukkan hobi Anda: ")

# Mencetak hasil ke layar
print("\n--- Data yang Anda Masukkan ---")
print("Nama  :", nama)
print("Umur  :", umur)
print("Hobi  :", hobi)
```
