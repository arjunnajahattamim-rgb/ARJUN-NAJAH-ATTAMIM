```python
# Dua "array" terpisah (pakai list sebagai array)
nama_barang = []   # array untuk nama
stok_barang = []   # array untuk stok (posisi sejajar dengan nama)

MAX_BARANG = 100

def cari_index(nama):
    for i, n in enumerate(nama_barang):
        if n == nama:
            return i
    return -1

def tambah_barang():
    if len(nama_barang) >= MAX_BARANG:
        print("Kapasitas penuh!")
        return
    nama = input("Nama barang  : ").strip()
    try:
        stok = int(input("Jumlah stok  : "))
    except ValueError:
        print("Stok harus angka.")
        return
    idx = cari_index(nama)
    if idx != -1:
        print("Sudah ada. Menambahkan stok ke barang lama...")
        stok_barang[idx] += stok
    else:
        nama_barang.append(nama)
        stok_barang.append(stok)
        print("Barang baru ditambahkan.")

def lihat_stok():
    if not nama_barang:
        print("Belum ada data.")
        return
    print(f"{'No':<4}{'Nama':<20}Stok")
    for i, (n, s) in enumerate(zip(nama_barang, stok_barang), start=1):
        print(f"{i:<4}{n:<20}{s}")

def perbarui_stok():
    nama = input("Nama barang yang diperbarui: ").strip()
    idx = cari_index(nama)
    if idx == -1:
        print("Barang tidak ditemukan.")
        return
    print("Stok lama:", stok_barang[idx])
    try:
        stok_baru = int(input("Stok baru: "))
    except ValueError:
        print("Stok harus angka.")
        return
    stok_barang[idx] = stok_baru
    print("Stok diperbarui.")

def cari_barang():
    nama = input("Nama barang yang dicari: ").strip()
    idx = cari_index(nama)
    if idx == -1:
        print("Tidak ditemukan.")
    else:
        print(f"Ditemukan: {nama_barang[idx]} | Stok: {stok_barang[idx]}")

def main():
    while True:
        print("\n=== Manajemen Stok Barang ===")
        print("1. Tambah barang")
        print("2. Lihat stok")
        print("3. Perbarui stok")
        print("4. Cari barang")
        print("0. Keluar")
        pilih = input("Pilih: ").strip()
        if pilih == "1": tambah_barang()
        elif pilih == "2": lihat_stok()
        elif pilih == "3": perbarui_stok()
        elif pilih == "4": cari_barang()
        elif pilih == "0":
            print("Terima kasih.")
            break
        else:
            print("Menu tidak valid.")

if __name__ == "__main__":
    main()
```
